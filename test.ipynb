```python
import math


def calculate_entropy(*args):
    total = sum(args)  
    entropy = 0

    for count in args:
        if count > 0:  
            probability = count / total
            entropy -= probability * math.log2(probability)

    return entropy

entropy = calculate_entropy(6, 2)
print(f"Entropy: {entropy:.4f}")
```
